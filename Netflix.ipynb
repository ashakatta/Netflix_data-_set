# 1. Data loading and basic checking

```python
# 1. load the data set into a frame
import pandas as pd
df = pd.read_csv('Netflix.csv')
print(df)
```

# 2. Data cleaning

```python

# check and report the missing values of the data set

df.isnull().sum()
```

```python
# Handling missing values
df1 = df.dropna()
print(df1.isnull().sum())
```

```python
# columns are converted into date time format using format='mixed' to handle whitespace and variations
df1['date_added'] = pd.to_datetime(df1['date_added'], format='mixed')
print(df1)
```

```python
# Analyze content type (Movie vs TV show) and their average release year wise
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(x='type', y='release_year', data=df1)
```

# 4. Trend and relationship exploration

```python
# use groupby() to identify the how many titles were added to Netflix in each year
df1.groupby(df1['date_added'].dt.year).size()
```

```python
# determine if there are any noticeble trend in the number of titles added per year
df1.groupby(df1['date_added'].dt.year).size().plot()
```

```python
# Analyze the relationship between release year and rating type visually.
plt.figure(figsize=(6,4))
sns.scatterplot(x="release_year", y="rating", data=df1)
plt.title("Release_year vs rating")
plt.show()
```

# 5. Visualizations


```python
# plot a count plot for type(Movies vs TV shows )
sns.countplot(x='type',data=df1)
```

```python
# plot a bar chart showing the top 5 countries by content type
top_5_countries = df1['country'].value_counts().head(5).index
df_top_5 = df1[df1['country'].isin(top_5_countries)]

sns.countplot(x='country', hue='type', data=df_top_5)
plt.title('Top 5 Countries by Content Type')
plt.xticks(rotation=45)
plt.show()
```

```python
# create a line chart showing yearly title addition
df1.groupby(df1['date_added'].dt.year).size().plot(kind='line')
```

```python
# plot a scatter plot for the release year vs duration (convrted to numeric if needed)
plt.figure(figsize=(6,4))
sns.scatterplot(x='release_year', y= 'duration',data=df1)
plt.title("Release_year vs duration")
plt.show()
```

```python
# generate a heat map of correlation of numeric(if applicale)
plt.figure(figsize=(6,4))
corr = df1.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
